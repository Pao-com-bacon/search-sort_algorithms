LINEAR SEARCH

- Searches item by item, checking each index and comparing it with our target value.
    - Basically an eliminatory search, without any data sorting.
- Computational cost grows linearly (O(n)).
    - Since it checks item by item, the number of search iterations depends directly on the number of elements in a list, array, or similar structure.

```python
import numpy as np

linear_data = np.arange(1, 1001, 2)
target_value = 11

def linear_search(lst, target):
    for i in range(len(lst)):  # Remember: negative checks stay outside the loop so the search doesn't stop prematurely
        if lst[i] == target:
            return i
    return -1  # -1 is a universal indicator for when a specific value is not found by a search algorithm

search_result = linear_search(linear_data, target_value)

if search_result != -1:
    print("=" * 100)
    print(f"The target number was found!\nAnswer Index: {search_result}")
    print("=" * 100)
else:
    print("=" * 100)
    print("The target number was NOT found!")
    print("=" * 100)
```

BINARY SEARCH

- Starts searching from the middle element of the list. If the middle element matches, the search is complete. If the middle element is greater, it searches the list to the left of the midpoint and splits that list in two. The same applies if the middle element is smaller.
    - This is called divide and conquer. We split the sorted list in the middle (median), compare it with our target, and separate it into the left side (smaller than the median) and the right side (greater than the median).

- Computational cost grows at a rate of O(log(n)).
    - This is highly efficient for searching. It occurs because an O(log(n)) algorithm has a ratio between the number of operations and the input size that decreases and approaches zero as n increases. In other words, the ratio O(log(n)) approaches zero as the value of n grows. That means for 100,000 data points, this ratio is 0.00005, which represents 0.005% of iterations relative to the total amount of data (50 iterations to find the target value).

```python
def binary_search (list, target):
    list.sort()
    first_i = 0
    last_i = len(list) - 1

    while first_i <= last_i: #until the index is bellow the last index, the code still keeps running. Searching the target value. Then, outside the while statment, the response is gonna be -1, following the convention used in this cases
        
        mid_i = ((first_i + last_i)//2) #the calculate of the middle index must be inside the while statment, because she is gonna be recalculating everytime until the target is found, or not. Otherwise, we going to be using an hybrid type of search, with binary and linear.
        mid_el = list[mid_i]
        
        if target == mid_el:
            return mid_i
        
        elif mid_el > target:
            last_i = mid_i - 1
            
        elif mid_el < target:
            first_i = mid_i + 1
        
    return -1 #the while statment keeps running, separating the data in two halfs, until find the target value. If not finds, then returns the -1.

binary_data = np.arange(1, 1001, 2)
target_binary = 78
func_result = binary_search(binary_data, target_binary)

if func_result != -1:
    print("="*100)
    print(f"The target value {target_binary} is in the index {str(func_result)}")
    print("="*100)
    
else:
    print("="*100)
    print(f"The target value {target_binary} is NOT in the list")
    print("="*100)
```
