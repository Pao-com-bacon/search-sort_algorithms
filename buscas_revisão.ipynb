BUSCA LINEAR

- Busca item a item, pegando cada um dos index e comparando com nosso valor alvo.
    - Basicamente uma busca eliminatória, sem qualquer classificação dos dados.
- Custo computacional cresce de maneira linear (O(n))
    - Como ele verifica item a item, o número de iterações para busca depende diretamente do número de elementos em uma lista, array e afins.

```python
import numpy as np

data = np.arange(1, 101, 2)
alvo_func = 11
print(data)

def busca_linear(lista, alvo):
        
    for i in range(len(lista)): #lembre se, testes contrários, ficam fora do for para não parar a busca antes
        if lista[i] == alvo:
            return i
    return -1 #-1 é um indicador universal para quando não achamos um determinado valor de buscas por algorítimo de busca.

resultado_busca = busca_linear(data, alvo_func)

if resultado_busca != -1:
    print("="*100)
    print(f"O número alvo foi encontrado!\nÍndice da Resposta: {resultado_busca}")
    print("="*100)
else:
    print("="*100)
    print("O número alvo NÃO foi encontrado!")
    print("="*100)
```

BUSCA BINÁRIA

- Busca a partir do elemento do meio da lista. Se o meio for igual, acabou. Caso o meio seja maior, ele vai procurar na lista à esquerda da metade e separar essa lista em dois. O mesmo valor para caso o do meio for maior.
    - Isso se chama dividir para conquistar. Dividimos a lista ordenada no meio (mediana), comparamos com nosso alvo e separamos entre lado esquerdo (menor que a mediana) e direita (maior que a mediana)

- Custo computacional cresce na forma de O(log(n)).
    - Isso é altamente eficiente em termos de busca. Isso ocorre porque um algorítimo O(log(⁡n)) possui uma relação entre o número de operações e o tamanho da entrada que diminui e tende a zero quando n aumenta. Ou seja, a razão entre entre O(log(n))/n se aproxima de zero, conforme valor de n aumenta. Ou seja, para 100 mil dados, essa razão é de 0,00005. O que significa 0,005 % de iterações sobre o número total de dados (50 iterações para achar o valor alvo)
